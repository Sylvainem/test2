```python
def split_bill(total, people):
    try:
        per_person = total / people
        print(f"每人應付:{per_person:.0f}元")
    except ZeroDivisionError:
        print("錯誤: 分攤人數不能為 0！")
    except:
        print("其它錯誤")


split_bill(600, "3")
split_bill(600, 0)
```
